```python
# =====================================================
# INSTALLAZIONE LIBRERIE (LangChain struttura nuova)
# =====================================================

!pip install -q langchain langchain-community langchain-text-splitters langchain-huggingface langchain-chroma sentence-transformers chromadb

# =====================================================
# IMPORT CORRETTI
# =====================================================

from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_text_splitters import CharacterTextSplitter

# =====================================================
# DOCUMENTI DI ESEMPIO (policy interne)
# =====================================================

docs = [
"""Loan approvals require credit risk assessment.
Customer income verification and identity verification are mandatory.
All lending operations must follow internal compliance guidelines.""",

"""Mortgage applications require evaluation of customer credit score,
loan-to-value ratio, and financial stability.
Risk management policies must be followed before approval."""
]

# =====================================================
# TEXT SPLITTING
# =====================================================

splitter = CharacterTextSplitter(chunk_size=300, chunk_overlap=50)
chunks = splitter.split_text("\n".join(docs))

# =====================================================
# EMBEDDINGS MODEL
# =====================================================

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# =====================================================
# VECTOR DATABASE (CHROMA)
# =====================================================

vectordb = Chroma.from_texts(chunks, embeddings)

# =====================================================
# RETRIEVER QUERY
# =====================================================

retriever = vectordb.as_retriever()

query = "What checks are required for loan approval?"

results = retriever.invoke(query)

# =====================================================
# OUTPUT
# =====================================================

for r in results:
    print(r.page_content)
    print("-----")
```
